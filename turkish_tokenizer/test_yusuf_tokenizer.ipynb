Bu notebook, `yusuf_tokenizer.py` dosyasındaki hem mevcut tokenizer fonksiyonlarını hem de yeni morfolojik decoder'ı kapsamlı şekilde test eder.

## Test Kapsamı:
1. **Temel Tokenization Tests** - Encode fonksiyonları

3. **Karşılaştırmalı Tests** - Eski vs Yeni decoder

```python
def test_basic_tokenization():
    """Temel tokenization testleri - Encode fonksiyonlarını test et"""
    
    print("🔤 TEMEL TOKENİZATION TESTLERİ")
    print("=" * 45)
    
    test_cases = [
        "merhaba dünya",
        "Ankara İstanbul", 
        "kitaplar evlerde",
        "YouTube Facebook",
        "çocuklarımızda",
        "Merhaba, nasılsın?",
        "satır1\nsatır2\ttab_test",
        "kitabımızda",
        "İstanbul'da",
        "öğrencilerimizin başarıları",
        "teknolojinin gelişmesi",
        "doktorlarımızla görüştük",
        "bilgisayarların programları",
        "müzisyenlerden öğrendiklerimiz",
        "Türkiye'nin en güzel şehirleri",
        "çocukluğumuzdan hatırladıklarımız",
        "arkadaşlarımızla yaptığımız seyahat",
        "öğretmenlerimizin verdiği ödevler",
        "yaşadığımız deneyimlerden",
        "gelecekte yapmak istediğimiz planlar",
        "ailelerimizle geçirdiğimiz zamanlar",
        "universitelerdeki öğrenci yaşamı",
        "şirketlerin çalışanlarına sunduğu imkanlar",
        "hastanelerde görev yapan hemşireler",
        "kütüphanelerde çalışan araştırmacılar",
        "mühendislerin tasarladığı projeler",
        "avukatların mahkemelerdeki savunmaları",
        "pazarlardaki satıcıların günlük hayatı",
        "öğrencilerin sınavlara hazırlanması"
    ]
    
    passed = 0
    total = len(test_cases)
    
    for i, text in enumerate(test_cases, 1):
        print(f"\n📝 Test {i}/{total}: '{text}'")
        
        try:
            # Yusuf tokenizer ile tokenize
            yt_result = yt.tokenize(text)
            
            # Eski tokenizer ile karşılaştır
            tt_result = tt.tokenize(text)
            
            print(f"   YT Tokens: {yt_result['tokens'][:5]}..." if len(yt_result['tokens']) > 5 else f"   YT Tokens: {yt_result['tokens']}")
            print(f"   YT IDs: {yt_result['ids'][:5]}..." if len(yt_result['ids']) > 5 else f"   YT IDs: {yt_result['ids']}")
            
            # Tokenization sonuçları aynı mı kontrol et
            if yt_result['tokens'] == tt_result['tokens'] and yt_result['ids'] == tt_result['ids']:
                print("   ✅ BAŞARILI: Tokenization eski sisteme eşit")
                passed += 1
            else:
                print("   ❌ FARKLI: Tokenization sonuçları farklı")
                print(f"   TT Tokens: {tt_result['tokens'][:5]}..." if len(tt_result['tokens']) > 5 else f"   TT Tokens: {tt_result['tokens']}")
                
        except Exception as e:
            print(f"   ❌ HATA: {str(e)}")
    
    print(f"\n🏁 Temel Tokenization Sonuçları: {passed}/{total} başarılı")
    return passed, total

# Temel tokenization testlerini çalıştır
basic_passed, basic_total = test_basic_tokenization()
```

```python
def test_comparative_decoders():
    """Eski ve yeni decoder'ları karşılaştır"""
    
    print("\n⚖️ KARŞILAŞTIRMALI DECODER TESTLERİ")
    print("=" * 50)
    
    test_cases = [
        "kitabı",
        "evlerde", 
        "çocuklarımızda",
        "teknolojisi",
        "gelmediler",
        "okuduğumuzda",
        "YouTube",
        "İstanbul'da",
        "öğrencilerimizin",
        "teknolojileri",
        "mühendislerden",
        "avukatlarımızla",
        "doktorlarımızın",
        "hemşirelerimize",
        "öğretmenlerimden",
        "pazarlardaki",
        "kütüphanelerde",
        "hastanelerde",
        "universitelerin",
        "şirketlerdeki",
        "çalışabilirsiniz",
        "yaşadıklarımızı",
        "düşünmediğimiz",
        "yürümekteydi",
        "yapabildiklerimiz",
        "öğrendiklerimizden",
        "hatırladıklarımız",
        "geliştirebileceklerimiz",
        "Facebook'tan",
        "WhatsApp'ta",
        "Instagram'da",
        "LinkedIn'de"
    ]
    
    print("Eski Turkish Tokenizer vs Yeni Morfolojik Decoder karşılaştırması:")
    
    results = {
        'old_better': 0,
        'new_better': 0,
        'equal': 0,
        'both_wrong': 0
    }
    
    for i, text in enumerate(test_cases, 1):
        print(f"\n📝 Test {i}/{len(test_cases)}: '{text}'")
        
        try:
            # Tokenize
            yt_result = yt.tokenize(text)
            token_ids = yt_result['ids']
            
            # Eski decoder
            start_time = time.time()
            old_decoded = tt.decode_text(token_ids)
            old_time = time.time() - start_time
            
            # Yeni morfolojik decoder 
            start_time = time.time()
            new_decoded = yt.morphological_decode(token_ids)
            new_time = time.time() - start_time
            
            print(f"   Orijinal:     '{text}'")
            print(f"   Eski decoder: '{old_decoded}' ({old_time:.4f}s)")
            print(f"   Yeni decoder: '{new_decoded}' ({new_time:.4f}s)")
            
            # Doğruluk karşılaştırması
            old_correct = (text == old_decoded)
            new_correct = (text == new_decoded)
            
            if old_correct and new_correct:
                print(f"   📊 Sonuç: ✅ Her ikisi de doğru")
                results['equal'] += 1
            elif old_correct and not new_correct:
                print(f"   📊 Sonuç: 🔙 Eski decoder daha iyi")
                results['old_better'] += 1
            elif not old_correct and new_correct:
                print(f"   📊 Sonuç: 🚀 Yeni decoder daha iyi")
                results['new_better'] += 1
            else:
                print(f"   📊 Sonuç: ❌ Her ikisi de yanlış")
                results['both_wrong'] += 1
                
        except Exception as e:
            print(f"   ❌ HATA: {str(e)}")
            results['both_wrong'] += 1
    
    print(f"\n🏁 KARŞILAŞTIRMA ÖZETİ:")
    print(f"   🚀 Yeni decoder daha iyi: {results['new_better']}")
    print(f"   🔙 Eski decoder daha iyi: {results['old_better']}")
    print(f"   ✅ Eşit performans: {results['equal']}")
    print(f"   ❌ Her ikisi de hatalı: {results['both_wrong']}")
    
    return results

# Karşılaştırmalı testleri çalıştır
comparison_results = test_comparative_decoders()
```

```python
def test_turkish_phonetic_rules():
    """Türkçe ses olayları kurallarını test et"""
    
    print("\n🎵 TÜRKÇE SES OLAYLARI TESTLERİ")
    print("=" * 45)
    
    # Ses olayları test grupları
    phonetic_tests = {
        "Ünsüz Yumuşaması": [
            ("kitap", "ı", "kitabı"),   # p→b
            ("ağaç", "ı", "ağacı"),     # ç→c  
            ("kanat", "ı", "kanadı"),   # t→d
            ("köpek", "i", "köpeği"),   # k→g
            # Yeni örnekler
            ("aşçı", "ya", "aşçıya"),   # ç→c
            ("tabak", "ı", "tabağı"),   # k→g
            ("yurt", "a", "yurda"),     # t→d
            ("sokak", "tan", "sokaktan"),# k→k (istisna)
            ("çiçek", "i", "çiçeği"),   # k→g
            ("dolap", "ı", "dolabı"),   # p→b
        ],
        "Ses Uyumu": [
            ("ev", "ler", "evler"),     # e→e (ön ünlü)
            ("masa", "lar", "masalar"), # a→a (art ünlü)
            ("kedi", "de", "kedide"),   # i→i (ön ünlü)
            ("araba", "da", "arabada"), # a→a (art ünlü)
            # Yeni örnekler
            ("öğrenci", "ler", "öğrenciler"), # e→e (ön ünlü)
            ("hastane", "de", "hastanede"),   # e→e (ön ünlü)
            ("teknoloji", "de", "teknolojide"), # i→i (ön ünlü)
            ("pazarlık", "tan", "pazarlıktan"), # ı→ı (art ünlü)
            ("mühendis", "ler", "mühendisler"), # i→i (ön ünlü)
            ("avukat", "lar", "avukatlar"),     # a→a (art ünlü)
        ],
        "Ünlü Düşmesi": [
            ("burun", "u", "burnu"),    # un→n
            ("karın", "ı", "karnı"),    # ın→n
            ("ağız", "ı", "ağzı"),      # ız→z
            # Yeni örnekler
            ("boyun", "u", "boynu"),    # un→n
            ("şehir", "i", "şehri"),    # ir→r
            ("beyin", "i", "beyni"),    # in→n
            ("göğüs", "ü", "göğsü"),    # üs→s
            ("alın", "ı", "alnı"),      # ın→n
            ("omuz", "u", "omzu"),      # uz→z
        ],
        "Kaynaştırma Ünsüzü": [
            ("su", "u", "suyu"),        # y eklenir
            ("anne", "i", "annesi"),    # s eklenir (3. şahıs)
            # Yeni örnekler
            ("kahve", "yi", "kahveyi"), # y eklenir
            ("okul", "a", "okula"),     # kaynaştırma yok
            ("baba", "sı", "babası"),   # s eklenir (3. şahıs)
            ("dede", "si", "dedesi"),   # s eklenir (3. şahıs)
            ("teyze", "yi", "teyzeyi"), # y eklenir
            ("çay", "ı", "çayı"),       # kaynaştırma yok
        ]
    }
    
    total_tests = 0
    total_passed = 0
    
    for rule_name, test_cases in phonetic_tests.items():
        print(f"\n🔍 {rule_name} Testleri:")
        print("-" * 30)
        
        rule_passed = 0
        
        for root, suffix, expected in test_cases:
            total_tests += 1
            
            try:
                # Tokenize et
                result = yt.tokenize(expected)
                
                # Morfolojik decode
                decoded = yt.morphological_decode(result['ids'])
                
                success = (expected == decoded)
                
                print(f"   {root} + {suffix} → {expected}")
                print(f"   Tokens: {result['tokens']}")
                print(f"   Decoded: '{decoded}'")
                
                if success:
                    print(f"   ✅ BAŞARILI")
                    rule_passed += 1
                    total_passed += 1
                else:
                    print(f"   ❌ BAŞARISIZ - Beklenen: '{expected}', Alınan: '{decoded}'")
                    
                print()
                
            except Exception as e:
                print(f"   ❌ HATA: {str(e)}")
                print()
        
        print(f"📊 {rule_name} Sonucu: {rule_passed}/{len(test_cases)} başarılı")
    
    print(f"\n🏁 GENEL SES OLAYLARI SONUCU: {total_passed}/{total_tests} başarılı")
    return total_passed, total_tests

# Ses olayları testlerini çalıştır
phonetic_passed, phonetic_total = test_turkish_phonetic_rules()
```

```python
def test_performance_comparison():
    """Performans karşılaştırması"""
    
    print("\n⚡ PERFORMANS KARŞILAŞTIRMASI")
    print("=" * 40)
    
    # Farklı boyutlarda test metinleri
    test_texts = {
        "Kısa (15 char)": "Merhaba dünya",
        "Orta (150 char)": "Bu bir Türkçe tokenizer test metnidir. Çeşitli kelimeler, ekler ve özel karakterler içerir. Morfolojik analiz yapılıyor." * 1,
        "Uzun (1500 char)": "Yapay zeka ve makine öğrenmesi alanında son yıllarda büyük ilerlemeler kaydedilmiştir. Bu ilerlemeler sayesinde bilgisayarlar insan gibi düşünebilir hale gelmiştir. Özellikle doğal dil işleme alanında tokenization çok önemli bir rol oynamaktadır." * 5,

        "Ağır Morfoloji": "öğrencilerimizin yapabildiklerimizden düşünmediğimiz geliştirebileceklerimiz çalışabilirsiniz",
        "Karışık İçerik": "Teknoloji şirketlerinde çalışan mühendislerimizin 2024 yılında @company #başarı ile geliştirdiği projeler",
        "Sosyal Medya": "@kullanici_adi #hashtag 😊 ile paylaştığı içerikler https://www.example.com sitesinde"
    }
    
    for size_name, text in test_texts.items():
        print(f"\n📊 {size_name} Metin Testi:")
        print(f"   Karakter sayısı: {len(text)}")
        
        try:
            # Tokenize (her iki sistem de aynı)
            start_time = time.time()
            result = yt.tokenize(text)
            tokenize_time = time.time() - start_time
            
            token_ids = result['ids']
            print(f"   Token sayısı: {len(token_ids)}")
            print(f"   ⏱️ Tokenize: {tokenize_time:.4f}s")
            
            # Eski decoder
            start_time = time.time()
            old_decoded = tt.decode_text(token_ids)
            old_decode_time = time.time() - start_time
            
            # Yeni morfolojik decoder
            start_time = time.time()
            new_decoded = yt.morphological_decode(token_ids)
            new_decode_time = time.time() - start_time
            
            print(f"   ⏱️ Eski decoder: {old_decode_time:.4f}s")
            print(f"   ⏱️ Yeni decoder: {new_decode_time:.4f}s")
            
            # Hız karşılaştırması
            if new_decode_time < old_decode_time:
                speedup = old_decode_time / new_decode_time
                print(f"   🚀 Yeni decoder {speedup:.2f}x daha hızlı")
            elif old_decode_time < new_decode_time:
                slowdown = new_decode_time / old_decode_time
                print(f"   🐌 Yeni decoder {slowdown:.2f}x daha yavaş")
            else:
                print(f"   ⚖️ Hızlar eşit")
            
            # Doğruluk karşılaştırması
            old_correct = (text.strip() == old_decoded.strip())
            new_correct = (text.strip() == new_decoded.strip())
            
            print(f"   ✅ Eski decoder doğru: {old_correct}")
            print(f"   ✅ Yeni decoder doğru: {new_correct}")
            
        except Exception as e:
            print(f"   ❌ HATA: {str(e)}")

# Performans testlerini çalıştır
test_performance_comparison()
```

```python
def test_edge_cases_yusuf():
    """Yusuf tokenizer için kenar durumları testleri"""
    
    print("\n🚨 KENAR DURUMLARI TESTLERİ (YUSUF TOKENIZER)")
    print("=" * 55)
    
    edge_cases = [
        {
            "name": "Boş String",
            "text": "",
            "test_decode": True
        },
        {
            "name": "Sadece Boşluk",
            "text": "   ",
            "test_decode": True
        },
        {
            "name": "Art Arda Büyük Harf",
            "text": "ABCD",
            "test_decode": True
        },
        {
            "name": "Türkçe Özel Karakterler",
            "text": "çğıöşüÇĞIÖŞÜ",
            "test_decode": True
        },
        {
            "name": "Karışık Noktalama",
            "text": "Merhaba!!! ...Nasılsın???",
            "test_decode": True
        },
        {
            "name": "Sadece Sayılar",
            "text": "123456789",
            "test_decode": True
        },
        {
            "name": "Karışık Whitespace",
            "text": "\\t\\n\\r  kelime  \\t\\n",
            "test_decode": True
        },
        # Yeni kenar durumları
        {
            "name": "Uzun Kelime Dizisi",
            "text": "çocuklarımızınkilerdenmiş",
            "test_decode": True
        },
        {
            "name": "Çok Uzun Ek Dizisi",
            "text": "yapabildiklerimizdenmiş",
            "test_decode": True
        },
        {
            "name": "Karışık Dil (İngilizce-Türkçe)",
            "text": "computer'da çalışıyoruz",
            "test_decode": True
        },
        {
            "name": "Sosyal Medya Hesapları",
            "text": "@kullanici_adi #hashtag",
            "test_decode": True
        },
        {
            "name": "Email ve URL",
            "text": "test@example.com ve https://www.site.com",
            "test_decode": True
        },
        {
            "name": "Sayılar ve Harfler",
            "text": "1. öğrenci 2. doktor 3. mühendis",
            "test_decode": True
        },
        {
            "name": "Emoji ve Özel Karakterler",
            "text": "Merhaba! 😊 Nasılsın? 🤔",
            "test_decode": True
        },
        {
            "name": "Çok Noktalama",
            "text": "!!!???...;;;:::---",
            "test_decode": True
        },
        {
            "name": "Parantez ve Köşeli Parantez",
            "text": "(kitap) [dergi] {gazete}",
            "test_decode": True
        },
        {
            "name": "Matematik İfadeleri",
            "text": "2+3=5 ve x²+y²=z²",
            "test_decode": True
        },
        {
            "name": "Para Birimleri",
            "text": "100₺ veya $50 yahut €30",
            "test_decode": True
        },
        {
            "name": "Kısaltmalar",
            "text": "Prof. Dr. Ahmet Bey'le",
            "test_decode": True
        },
        {
            "name": "Çok Satırlı Metin",
            "text": "Birinci satır\\nİkinci satır\\nÜçüncü satır",
            "test_decode": True
        },
        {
            "name": "Tab ve Karışık Boşluklar",
            "text": "kelime\\t\\tbaşka_kelime   son_kelime",
            "test_decode": True
        }
    ]
    
    passed = 0
    total = len(edge_cases)
    
    for i, test_case in enumerate(edge_cases, 1):
        print(f"\\n🔍 Test {i}/{total}: {test_case['name']}")
        print(f"   Giriş: {repr(test_case['text'])}")
        
        try:
            # Tokenize
            result = yt.tokenize(test_case['text'])
            print(f"   Tokens: {result['tokens']}")
            print(f"   Token sayısı: {len(result['tokens'])}")
            
            if test_case['test_decode']:
                # Eski decoder
                try:
                    old_decoded = tt.decode_text(result['ids'])
                    print(f"   Eski decode: {repr(old_decoded)}")
                except Exception as e:
                    print(f"   Eski decode HATA: {str(e)}")
                
                # Yeni decoder
                try:
                    new_decoded = yt.morphological_decode(result['ids'])
                    print(f"   Yeni decode: {repr(new_decoded)}")
                    
                    # Roundtrip kontrolü
                    if test_case['text'] == new_decoded:
                        print(f"   ✅ BAŞARILI: Roundtrip doğru")
                        passed += 1
                    else:
                        print(f"   ⚠️ FARKLI: Roundtrip farklı")
                        print(f"      Orijinal: {repr(test_case['text'])}")
                        print(f"      Decoded:  {repr(new_decoded)}")
                        
                except Exception as e:
                    print(f"   ❌ Yeni decode HATA: {str(e)}")
            else:
                print(f"   ✅ BAŞARILI: Tokenization tamamlandı")
                passed += 1
                
        except Exception as e:
            print(f"   ❌ TOKENIZE HATA: {str(e)}")
    
    print(f"\\n🏁 Kenar Durumları Sonucu: {passed}/{total} başarılı")
    return passed, total

# Kenar durumları testlerini çalıştır
edge_passed, edge_total = test_edge_cases_yusuf()
```

```python
def final_summary():
    """Tüm testlerin genel özeti"""
    
    print("\n" + "=" * 60)
    print("🏆 YUSUF TOKENIZER TEST ÖZETİ")
    print("=" * 60)
    
    # Test sonuçlarını topla
    total_tests = 0
    total_passed = 0
    
    if 'basic_passed' in globals():
        total_tests += basic_total
        total_passed += basic_passed
        print(f"📝 Temel Tokenization: {basic_passed}/{basic_total} ({'✅' if basic_passed == basic_total else '⚠️'})")
    
    if 'morph_passed' in globals():
        total_tests += morph_total
        total_passed += morph_passed
        print(f"🧬 Morfolojik Decoder: {morph_passed}/{morph_total} ({'✅' if morph_passed == morph_total else '⚠️'})")
    
    if 'phonetic_passed' in globals():
        total_tests += phonetic_total
        total_passed += phonetic_passed
        print(f"🎵 Ses Olayları: {phonetic_passed}/{phonetic_total} ({'✅' if phonetic_passed == phonetic_total else '⚠️'})")
    
    if 'edge_passed' in globals():
        total_tests += edge_total
        total_passed += edge_passed
        print(f"🚨 Kenar Durumları: {edge_passed}/{edge_total} ({'✅' if edge_passed == edge_total else '⚠️'})")
    
    if 'comparison_results' in globals():
        new_better = comparison_results['new_better']
        old_better = comparison_results['old_better']
        equal = comparison_results['equal']
        total_comp = new_better + old_better + equal + comparison_results['both_wrong']
        print(f"⚖️ Karşılaştırma: Yeni {new_better}, Eski {old_better}, Eşit {equal} / {total_comp}")
    
    print("\n" + "-" * 60)
    
    if total_tests > 0:
        success_rate = (total_passed / total_tests) * 100
        print(f"📊 GENEL BAŞARI ORANI: {total_passed}/{total_tests} ({success_rate:.1f}%)")
        
        if success_rate >= 90:
            print("🏆 MÜKEMMELDİR! Tokenizer çok iyi çalışıyor.")
        elif success_rate >= 75:
            print("✅ İYİDİR! Küçük iyileştirmeler yapılabilir.")
        elif success_rate >= 50:
            print("⚠️ ORTA! Bazı problemler var, iyileştirme gerekli.")
        else:
            print("❌ ZAYIF! Ciddi problemler var, büyük revizyonlar gerekli.")
    
    print("\n" + "=" * 60)

# Final özeti göster
final_summary()
```

```python
def debug_tokenization_issues():
    """Debug specific tokenization problems"""
    
    print("🔧 DEBUG: Problem Tokenization Cases")
    print("=" * 50)
    
    problem_cases = [
        'yürümekteydi',
        'çalışabilirsiniz', 
        'tabağı',
        'çiçeği',
        'düşünmediğimiz',
        'yapabildiklerimiz'
    ]
    
    for i, text in enumerate(problem_cases, 1):
        print(f"\n🔍 Debug {i}/{len(problem_cases)}: '{text}'")
        
        try:
            # Tokenize
            result = yt.tokenize(text)
            print(f"   Tokens: {result['tokens']}")
            print(f"   IDs: {result['ids']}")
            
            # Check what forms are available for each token ID
            for j, token_id in enumerate(result['ids']):
                if token_id in yt.reverse_dict:
                    forms = yt.reverse_dict[token_id]
                    if len(forms) > 1:
                        print(f"   Token {j} ID {token_id} has forms: {forms}")
            
            # Decode
            decoded = yt.morphological_decode(result['ids'])
            print(f"   Decoded: '{decoded}'")
            print(f"   Match: {'✅' if text == decoded else '❌'}")
            
        except Exception as e:
            print(f"   ERROR: {str(e)}")

# Run debug
debug_tokenization_issues()
```
