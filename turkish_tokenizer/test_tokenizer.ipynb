```python
# Turkish Tokenizer Test Suite
import turkish_tokenizer as tt
import json
import time

def test_tokenizer_comprehensive():
    """Kapsamlı tokenizer test fonksiyonu"""
    
    print("🔥 TURKISH TOKENIZER TEST SÜİTİ 🔥")
    print("=" * 50)
    
    test_cases = [
        # Temel test case'leri
        {
            "name": "Basit Kelimeler",
            "text": "merhaba dünya",
            "expected_issues": []
        },
        {
            "name": "Büyük Harf Testi",
            "text": "Ankara İstanbul",
            "expected_issues": []
        },
        {
            "name": "Karışık Büyük Harf",
            "text": "YouTube Facebook Twitter",
            "expected_issues": []
        },
        {
            "name": "Ekli Kelimeler",
            "text": "kitaplar evlerde çocuklarımız",
            "expected_issues": []
        },
        {
            "name": "Noktalama İşaretleri",
            "text": "Merhaba, nasılsın? İyi misin!",
            "expected_issues": []
        },
        {
            "name": "Whitespace Karakterleri",
            "text": "satır1\nsatır2\ttab_test",
            "expected_issues": []
        },
        {
            "name": "Art Arda Büyük Harf",
            "text": "ABC DEF GHI",
            "expected_issues": []
        },
        {
            "name": "Ses Uyumu Testi",
            "text": "evde evlerde okumak okuyor",
            "expected_issues": []
        },
        {
            "name": "Uzun Metin",
            "text": """Yapay zeka teknolojisi günümüzde hızla gelişmekte. 
            Bu gelişmeler sayesinde bilgisayarlar insan gibi düşünebiliyor.""",
            "expected_issues": []
        }
    ]
    
    total_tests = len(test_cases)
    passed_tests = 0
    
    for i, test_case in enumerate(test_cases, 1):
        print(f"\n📝 Test {i}/{total_tests}: {test_case['name']}")
        print(f"Giriş: '{test_case['text']}'")
        
        try:
            # Tokenize et
            start_time = time.time()
            result = tt.tokenize(test_case['text'])
            tokenize_time = time.time() - start_time
            
            # Decode et
            start_time = time.time()
            decoded = tt.decode_text(result['ids'])
            decode_time = time.time() - start_time
            
            # Sonuçları göster
            print(f"Tokens: {result['tokens'][:10]}..." if len(result['tokens']) > 10 else f"Tokens: {result['tokens']}")
            print(f"IDs: {result['ids'][:10]}..." if len(result['ids']) > 10 else f"IDs: {result['ids']}")
            print(f"Decoded: '{decoded}'")
            print(f"⏱️ Tokenize: {tokenize_time:.4f}s, Decode: {decode_time:.4f}s")
            
            # Başarı kontrolü (basit karşılaştırma)
            if decoded.strip() == test_case['text'].strip():
                print("✅ BAŞARILI: Decode sonucu orijinal metinle eşleşiyor")
                passed_tests += 1
            else:
                print("❌ BAŞARISIZ: Decode sonucu farklı")
                print(f"Beklenen: '{test_case['text']}'")
                print(f"Alınan:   '{decoded}'")
                
        except Exception as e:
            print(f"❌ HATA: {str(e)}")
    
    print(f"\n🏁 TEST SONUÇLARI: {passed_tests}/{total_tests} başarılı")
    return passed_tests == total_tests

# Testleri çalıştır
test_tokenizer_comprehensive()
```

```python
def test_edge_cases():
    """Kenar durumları test fonksiyonu"""
    
    print("\n🚨 KENAR DURUMLARI TESTLERİ 🚨")
    print("=" * 40)
    
    edge_cases = [
        {
            "name": "Boş String",
            "text": "",
            "should_work": True
        },
        {
            "name": "Sadece Boşluk",
            "text": "   ",
            "should_work": True
        },
        {
            "name": "Sadece Büyük Harf Token",
            "text": "A",
            "should_work": True
        },
        {
            "name": "Art Arda Büyük Harf",
            "text": "AAAA",
            "should_work": True
        },
        {
            "name": "Özel Karakterler",
            "text": "çğıöşü",
            "should_work": True
        },
        {
            "name": "Sayılar",
            "text": "123 456",
            "should_work": True
        },
        {
            "name": "Karışık Özel Durumlar",
            "text": "\t\n  ABC\t\nxyz\n",
            "should_work": True
        }
    ]
    
    print("Bu testler düzeltmelerimizi doğrulamak için:")
    
    for test_case in edge_cases:
        print(f"\n🔍 {test_case['name']}: '{repr(test_case['text'])}'")
        
        try:
            # Tokenize
            result = tt.tokenize(test_case['text'])
            
            # Decode
            decoded = tt.decode_text(result['ids'])
            
            print(f"✅ Başarılı - Tokens: {len(result['tokens'])}, IDs: {len(result['ids'])}")
            print(f"   Tokens: {result['tokens']}")
            print(f"   Decoded: {repr(decoded)}")
            
        except Exception as e:
            print(f"❌ Hata: {str(e)}")

def test_decode_error_cases():
    """Decode fonksiyonundaki hata durumlarını test et"""
    
    print("\n🛡️ DECODE HATA DURUMLARI TESTLERİ 🛡️")
    print("=" * 45)
    
    # Test case 1: Boş liste
    print("Test 1: Boş liste")
    try:
        result = tt.decode_text([])
        print(f"✅ Başarılı: '{result}'")
    except Exception as e:
        print(f"❌ Hata: {e}")
    
    # Test case 2: Bilinmeyen ID
    print("\nTest 2: Bilinmeyen ID")
    try:
        result = tt.decode_text([99999])
        print(f"✅ Başarılı (uyarı bekleniyor): '{result}'")
    except Exception as e:
        print(f"❌ Hata: {e}")
    
    # Test case 3: Art arda uppercase
    print("\nTest 3: Art arda uppercase tokens")
    try:
        # 0 = <uppercase>, varsayalım ki 385 = "test" gibi bir kelime
        result = tt.decode_text([0, 0, 385] if 385 in tt.reverse_dict else [0, 0, 1])
        print(f"✅ Başarılı: '{result}'")
    except Exception as e:
        print(f"❌ Hata: {e}")

# Kenar durumu testlerini çalıştır
test_edge_cases()
test_decode_error_cases()
```

```python
def performance_test():
    """Performans testi"""
    
    print("\n⚡ PERFORMANS TESTLERİ ⚡")
    print("=" * 30)
    
    # Farklı boyutlarda metinler
    test_texts = {
        "Kısa": "Merhaba dünya",
        "Orta": "Bu bir Türkçe tokenizer test metnidir. Çeşitli kelimeler ve ekler içerir." * 5,
        "Uzun": "Yapay zeka ve makine öğrenmesi alanında son yıllarda büyük ilerlemeler kaydedilmiştir. " * 50
    }
    
    for name, text in test_texts.items():
        print(f"\n📊 {name} metin testi:")
        print(f"   Karakter sayısı: {len(text)}")
        
        # Tokenize performansı
        start_time = time.time()
        result = tt.tokenize(text)
        tokenize_time = time.time() - start_time
        
        # Decode performansı
        start_time = time.time()
        decoded = tt.decode_text(result['ids'])
        decode_time = time.time() - start_time
        
        print(f"   Token sayısı: {len(result['tokens'])}")
        print(f"   ⏱️ Tokenize: {tokenize_time:.4f}s")
        print(f"   ⏱️ Decode: {decode_time:.4f}s")
        print(f"   🔄 Roundtrip başarılı: {decoded.strip() == text.strip()}")

def analyze_tokenization(text):
    """Detaylı tokenization analizi"""
    
    print(f"\n🔬 DETAYLI ANALİZ: '{text}'")
    print("=" * 50)
    
    result = tt.tokenize(text)
    decoded = tt.decode_text(result['ids'])
    
    print(f"Orijinal: '{text}'")
    print(f"Decoded:  '{decoded}'")
    print(f"Eşleşme: {text == decoded}")
    
    print(f"\nToken Detayları:")
    for i, (token, id_val) in enumerate(zip(result['tokens'], result['ids'])):
        token_type = "SPECIAL" if id_val in [0,1,2,3,4] else "WORD/BPE"
        print(f"  {i:2d}: '{token:15s}' -> ID: {id_val:5d} ({token_type})")
    
    # Token türü analizi
    special_count = sum(1 for id_val in result['ids'] if id_val in [0,1,2,3,4])
    root_count = sum(1 for id_val in result['ids'] if id_val in tt.roots.values())
    suffix_count = sum(1 for id_val in result['ids'] if id_val in tt.suffixes.values())
    bpe_count = len(result['ids']) - special_count - root_count - suffix_count
    
    print(f"\nToken Türü Dağılımı:")
    print(f"  🔹 Özel tokenlar: {special_count}")
    print(f"  🌿 Kökler: {root_count}")
    print(f"  📎 Ekler: {suffix_count}")
    print(f"  🧩 BPE tokenlar: {bpe_count}")
    print(f"  📊 Toplam: {len(result['ids'])}")

# Performans testlerini çalıştır
performance_test()

# Örnek detaylı analiz
analyze_tokenization("kitaplarımızda")
analyze_tokenization("YouTube'da")
```

```python
def interactive_test():
    """İnteraktif test fonksiyonu - Jupyter'de istediğiniz metni test edebilirsiniz"""
    
    def test_text(text):
        """Verilen metni test et ve sonuçları göster"""
        print(f"🔤 Test Metni: '{text}'")
        print("-" * 50)
        
        try:
            # Tokenize
            result = tt.tokenize(text)
            
            # Decode
            decoded = tt.decode_text(result['ids'])
            
            # Sonuçları göster
            print(f"📝 Tokenlar: {result['tokens']}")
            print(f"🔢 ID'ler: {result['ids']}")
            print(f"🔄 Decode Sonucu: '{decoded}'")
            print(f"✅ Başarı Durumu: {text == decoded}")
            
            if text != decoded:
                print(f"⚠️ Fark tespit edildi!")
                print(f"   Orijinal: '{text}'")
                print(f"   Decode:   '{decoded}'")
            
            return result, decoded
            
        except Exception as e:
            print(f"❌ Hata: {str(e)}")
            return None, None
    
    return test_text

# İnteraktif test fonksiyonunu hazırla
test_function = interactive_test()

# Hızlı test için örnek kullanım:
test_function("istediğiniz metin")
```

```python
def test_turkish_case_handling():
    """Türkçe büyük/küçük harf işleme testleri"""
    
    print("\n🇹🇷 TÜRKÇE BÜYÜK HARF TESTLERİ 🇹🇷")
    print("=" * 40)
    
    # Türkçe'ye özgü problemli kelimeler
    turkish_test_cases = [
        "İstanbul",
        "İzmir", 
        "İnsan",
        "İçecek",
        "İlginç",
        "Şehir",
        "Çocuk",
        "Ğünümüz",
        "Öğretmen",
        "Üstad"
    ]
    
    print("Bu test, Türkçe karakterlerin doğru işlendiğini kontrol eder:")
    
    passed = 0
    total = len(turkish_test_cases)
    
    for word in turkish_test_cases:
        print(f"\n🔤 '{word}':")
        
        try:
            result = tt.tokenize(word)
            decoded = tt.decode_text(result['ids'])
            
            success = word == decoded
            print(f"   Tokens: {result['tokens']}")
            print(f"   Decoded: '{decoded}'")
            print(f"   {'✅ BAŞARILI' if success else '❌ BAŞARISIZ'}")
            
            if success:
                passed += 1
            else:
                print(f"   Beklenen: '{word}'")
                print(f"   Alınan:   '{decoded}'")
                
        except Exception as e:
            print(f"   ❌ HATA: {str(e)}")
    
    print(f"\n📊 Türkçe Büyük Harf Sonuçları: {passed}/{total} başarılı")
    return passed == total

# Türkçe harf testini çalıştır
test_turkish_case_handling()
```
