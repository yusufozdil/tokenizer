```python
from datasets import load_dataset

dswiki = load_dataset("wikimedia/wikipedia", "20231101.tr")
dfwiki = dswiki['train'].to_pandas()
dfwiki
```

```python
texts_with_title_wiki = dfwiki['title'] + " " + dfwiki['text']
texts_wiki = texts_with_title_wiki.to_list()
texts_wiki
```

```python
texts_lower_case_wiki = [text.lower() for text in texts_wiki]
```

```python
dsmasallar = load_dataset("umutphp/masallar")
dfmasallar = dsmasallar['train'].to_pandas()
dfmasallar
```

```python
texts_with_title_masallar = dfmasallar['title'] + " " + dfmasallar['text']
texts_masallar = texts_with_title_masallar.to_list()
texts_masallar
```

```python
texts_lower_case_masallar = [text.lower() for text in texts_masallar]
```

```python
dsnews = load_dataset("habanoz/news-tr-1.8M")
dfnews = dsnews['train'].to_pandas()
dfnews
```

```python
texts_with_title_news = dfnews['text']
texts_news = texts_with_title_news.to_list()
texts_news
```

```python
texts_lower_case_news = [text.lower() for text in texts_news]
```

```python
texts_all = texts_lower_case_wiki + texts_lower_case_masallar + texts_lower_case_news
```

```python
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace

tokenizer = Tokenizer(BPE())
tokenizer.pre_tokenizer = Whitespace()
tokenizer
```

```python
from tokenizers import Tokenizer
tokenizer.train_from_iterator(texts_all, trainer=trainer)
tokenizer.save("new_custom_bpe_tokenizer_yusuf.json")
tokenizer.get_vocab_size()
```
